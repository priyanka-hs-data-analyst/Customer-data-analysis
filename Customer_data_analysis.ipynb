```python
df['Total_Sales']=df['quantity']*df['price']
print(df)
```

```python
df['invoice_date']=pd.to_datetime(df['invoice_date'])
df['year']=df['invoice_date'].dt.year
df['month']=df['invoice_date'].dt.month
print(df)
```

```python
#Gender distribution
df['gender'].value_counts()
```

```python
#Category distribution
df['category'].value_counts()
```

```python
#Payment method distribution
df['payment_method'].value_counts()
```

```python
#Total sales
round(df['Total_Sales'].sum(),2)
```

```python
#Sales by category
df.groupby('category')['Total_Sales'].sum().sort_values(ascending=False)
```

```python
#Sales by shopping mall
df.groupby('shopping_mall')['Total_Sales'].sum().sort_values()
```

```python
#Sales by gender
gen_sales=df.groupby('gender')['Total_Sales'].sum()
gen_sales.apply(lambda x: f"{x:,.2f}")
```

```python
#Sales by year
yr_sales=df.groupby('year')['Total_Sales'].sum()
yr_sales.apply(lambda x:f"{x:,.2f}")
```

```python
#Sales by month
df.groupby('month')['Total_Sales'].sum()
```

```python
#Most sold categories
df.groupby('category')['quantity'].sum().sort_values(ascending=False)
```

```python
#Category vs gender
df.groupby(['category','gender'])['Total_Sales'].sum()
```

```python
#Payment method vs sales
pm_sales=df.groupby('payment_method')['Total_Sales'].sum().sort_values(ascending=False)
pm_sales.apply(lambda x:f"{x:,.2f}")
```

```python
#Task1 1 - How is the shopping distribution according to gender?
#based on count
df['gender'].value_counts()

#df['gender'].value_counts(normalize=True)*100
```

```python
#Task 2 - Which gender did we sell more products to?
df.groupby('gender')['quantity'].sum()
```

```python
#Task 3 - Which gender generated more revenue?
gen_sales=df.groupby('gender')['Total_Sales'].sum()
gen_sales.apply(lambda x:f"{x:,.2f}")
```

```python
#Task 4 - Distribution of purchase categories relative to other columns
#Category vs gender
pd.crosstab(df['category'],df['gender'])

#category vs payment method
pd.crosstab(df['category'],df['payment_method'])

#category vs shopping mall
pd.crosstab(df['category'],df['shopping_mall'])

#category vs sales
cat_sales=df.groupby('category')['Total_Sales'].sum().sort_values(ascending=False)
cat_sales.apply(lambda x:f"{x:,.2f}")

#Category vs quantity
df.groupby('category')['quantity'].sum()

#category vs age group
df['age_group']=pd.cut(df['age'],bins=[18,25,35,50,65],labels=['18-25','26-35','36-50','50+'])
pd.crosstab(df['category'],df['age_group'])
```

```python
#Task 5 - How is the shopping distribution according to age?
#number of purchses by age group
df['age_group'].value_counts().sort_values(ascending=False)

#Sales by age group
df.groupby('age_group',observed=True)['Total_Sales'].sum().sort_values(ascending=False)

#Quantity by age group
df.groupby('age_group',observed=True)['Total_Sales'].sum()
```

```python
#Task 6 - Which age cat did we sell more products to?
df.groupby('age_group',observed=True)['quantity'].sum().sort_values(ascending=False)
```

```python
#Task 7 - Which age cat generated more revenue?
df.groupby('age_group',observed=True)['Total_Sales'].sum().sort_values(ascending=False)
```

```python
#Task 8 - Does the payment method have a relation with other columns?
#Payment method distribution
df['payment_method'].value_counts()

#payment method vs gender
pd.crosstab(df['gender'],df['payment_method'],normalize='columns')*100

#payment method vs category
pd.crosstab(df['category'],df['payment_method'])

#payment method vs age group
pd.crosstab(df['age_group'],df['payment_method'])

#payment method vs sales
pay_sales=df.groupby('payment_method')['Total_Sales'].sum().sort_values(ascending=False)
pay_sales.apply(lambda x:f"{x:,.2f}")

#payment method vs shopping mall
pd.crosstab(df['shopping_mall'],df['payment_method'])
```
